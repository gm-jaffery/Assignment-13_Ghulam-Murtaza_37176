```python
import pandas as pd 
# Load dataset 
df = pd.read\_csv("teaching\_prices\_clean.csv") 
# 1\. Calculate overall mean and median per item 
stats = df.groupby('item')['price'].agg(['mean', 'median']).reset\_index() 
# 2\. Calculate Jan 01 and Mar 19 averages &amp; percentage changes 
jan1 = df[df['date'] == '2026-01-01'].groupby('item')['price'].mean() 
mar19 = df[df['date'] == '2026-03-19'].groupby('item')['price'].mean() 
stats['jan1\_avg'] = stats['item'].map(jan1) 
stats['mar19\_avg'] = stats['item'].map(mar19) 
stats['pct\_change'] = ((stats['mar19\_avg'] - stats['jan1\_avg']) / stats['jan1\_avg']) \* 100 
print("=== SUMMARY STATISTICS ===") 
print(stats.to\_string(index=False)) 
# 3\. Outlier Impact Comparison (Staple A) 
staple\_a\_clean = df[df['item'] == 'Staple A']['price'].tolist() 
staple\_a\_messy = staple\_a\_clean + [1050.0] 
clean\_series = pd.Series(staple\_a\_clean) 
messy\_series = pd.Series(staple\_a\_messy) 
print("\\n=== STAPLE A OUTLIER AUDIT ===") 
print(f"Clean -&gt; Mean: {clean\_series.mean():.2f}, Median: {clean\_series.median():.2f}") print(f"Messy -&gt; Mean: {messy\_series.mean():.2f}, Median: {messy\_series.median():.2f}")
```
